```python
def factorial(x):
    n=1
    for i in range(1,x+1):
        n*=i
    return n

factorial(5)    
```

```python
def prime(n):
    if(n<2):
        return False
    if(n==2):
        return True
    for i in range(2, int(n ** 0.5)+1):
        if(n%i==0):
            return False
    return True


        
    
```

```python
num =7
if prime(num):
    print(f"{num} is prime number")
else:
    print(f"{num} is not prime number")
```
